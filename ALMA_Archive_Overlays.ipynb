```python
rslt = Alma.query_region(SkyCoord(359*u.deg, 0*u.deg, frame='galactic'), radius=1*u.deg)
len(rslt)
```

```python
crs = SkyCoord(rslt['s_ra'], rslt['s_dec'], frame='icrs')
```

```python
import operator

def _regionlist_to_single_region(region_list):
    """
    Recursively merge a region list into a single compound region
    """
    import regions
    if len(region_list) == 1:
        return region_list[0]
    left = _regionlist_to_single_region(region_list[:len(region_list)//2])
    right = _regionlist_to_single_region(region_list[len(region_list)//2:])
    return regions.CompoundSkyRegion(left, right, operator.or_)


def parse_region(reg):
    spl = reg.split()
    if spl[0] == 'Polygon':
        csys = spl[1]
        points = SkyCoord([(float(spl[ii]), float(spl[ii+1]))*u.deg
                  for ii in range(2, len(spl), 2)], frame=csys.lower())
        return regions.PolygonSkyRegion(points)
    elif spl[0] == 'Circle':
        csys = spl[1]
        points = SkyCoord(float(spl[2])*u.deg, float(spl[3])*u.deg, frame=csys.lower())
        return regions.CircleSkyRegion(points, radius=float(spl[4])*u.deg)
    elif spl[0] == 'Union':
        csys = spl[1].lower()
        # Hack
        regs = reg.split("Polygon")[1:]
        polys = []
        for row in regs:
            spl_ = row.strip(")").split()
            points = SkyCoord([(float(spl_[ii]), float(spl_[ii+1]))*u.deg
                      for ii in range(2, len(spl_), 2)], frame=csys.lower())
            poly = regions.PolygonSkyRegion(points)
            polys.append(poly)
        return _regionlist_to_single_region(polys)

    else:
        print(reg)
        raise ValueError(reg)
```

```python
regs = regions.Regions([parse_region(x) for x in rslt['s_region']])
len(regs)
```

```python
compound_regs = _regionlist_to_single_region(regs)
```

```python
from astropy.io import fits
```

```python
ww = wcs.WCS(fits.Header.fromstring("""\
SIMPLE  =                    T
BITPIX  =                  -64
NAXIS   =                    2
NAXIS1  =                 1000
NAXIS2  =                 1000
WCSAXES =                    2
CRPIX1  =                500.0
CRPIX2  =                500.0
CDELT1  =    -0.002
CDELT2  =     0.002
CUNIT1  = 'deg'               
CUNIT2  = 'deg'               
CTYPE1  = 'GLON-TAN'          
CTYPE2  = 'GLAT-TAN'          
CRVAL1  =                359.0
CRVAL2  =                  0.0
MJDREF  =                  0.0
""", sep='\n'))
```

```python
pregs = regions.Regions([reg.to_pixel(ww) for reg in regs])
```

```python
compound_pregs = compound_regs.to_pixel(ww)
```
